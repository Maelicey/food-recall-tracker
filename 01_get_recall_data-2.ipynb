   # Milestone 1: Get the data

```python
# Load the tools we need
import requests
import pandas as pd
import zipfile
import io
import json
```

```python
# Download the full food recall dataset from openFDA (zipped)
zip_url = "https://download.open.fda.gov/food/enforcement/food-enforcement-0001-of-0001.json.zip"
zip_response = requests.get(zip_url)
print(zip_response.status_code)
```

```python
# Unzip the file and load the recalls into a table
zip_file = zipfile.ZipFile(io.BytesIO(zip_response.content))
file_name = zip_file.namelist()[0]
full_data = json.loads(zip_file.read(file_name))

df_full = pd.DataFrame(full_data["results"])
print(df_full.shape)
```

```python
# Keep only the 10 columns useful for the recall tracker
features = [
    "recall_number",
    "event_id",
    "report_date",
    "recall_initiation_date",
    "recalling_firm",
    "product_description",
    "reason_for_recall",
    "classification",
    "status",
    "distribution_pattern",
]

df_full_clean = df_full[features].copy()
print(df_full_clean.shape)
df_full_clean.head()
```

```python
# Save the cleaned dataset as a CSV file
df_full_clean.to_csv("food_recalls_clean.csv", index=False)
```

# Milestone 2: Clean the data

```python
# Convert report_date from text to a real date
df_full_clean["report_date"] = pd.to_datetime(df_full_clean["report_date"], format="%Y%m%d")

df_full_clean["report_date"].head()
```

```python
#convert recall initiation date from text to a real date
df_full_clean["recall_initiation_date"] = pd.to_datetime(df_full_clean["recall_initiation_date"], format="%Y%m%d", errors = "coerce")

df_full_clean["recall_initiation_date"].head()
```
